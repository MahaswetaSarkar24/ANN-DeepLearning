```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
# n_samples=100 → A total of 100 data points (rows) will be generated.
# n_features=2 → Each data point will have 2 features (columns).
# n_informative=1 → Out of the 2 features, 1 feature will be informative for determining the class.
# n_redundant=0 → No additional or derived features will be generated.
# n_classes=2 → There will be 2 classes (e.g., 0 and 1).
# n_clusters_per_class=1 → The data points of each class will form 1 cluster/group.
# random_state=41 → A seed value to ensure that the same dataset is generated every time the code is run.
# hypercube=False → The informative features will be generated without following the hypercube structure.
# class_sep=10 → The separation between the two classes will be very large, making them easier to distinguish.
```

```python
def step(Z):
    return 1 if Z>0 else 0
    
def perceptron_model(X,y):
    X = np.insert(X,0,1,axis=1)
    Weights = np.ones(X.shape[1])
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],Weights))
        Weights = Weights + (lr * (y[j]-y_hat) * X[j])
    return Weights[0],Weights[1:]
```

```python
bias,weights = perceptron_model(X,y)
print(f'Bias is: {bias}')
print(f'Weights are: {weights}')
```
